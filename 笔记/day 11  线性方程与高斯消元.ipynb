## 高斯消元法求解线性方程组

### 高斯消元复杂度是$O(n^{3})$，方程组做行变换，数值更稳定，误差更小
标准化、矩阵化、算法化

Example：$$
\begin{align}
2x + y - z &= 8 \\
-3x - y + 2z &= -11 \\
-2x + y + 2z &= -3
\end{align}
$$

简化书写，常把系数和常数提取出来

增广矩阵：$$
\left[
\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
-3 & -1 & 2 & -11 \\
-2 & 1 & 2 & -3
\end{array}
\right]
$$
高斯消元的三大变换：
1. 交换变行
2. 某一行乘以非0常数
3. 某一行加上另一行倍数

目的：把矩阵变成上三角矩阵，从上往下代回求解


$$\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 1/2 & 1/2 & 1 \\
0 & 0 & -1 & 1
\end{array}\right]$$


---

## 第一步：消去第一列下方的元素

**目标**：让第 2 行和第 3 行的第一列变成 0。

**变换操作**：
- $R_2 \leftarrow R_2 + \frac{3}{2}R_1$
- $R_3 \leftarrow R_3 + R_1$

**变换后的矩阵**：

$$
\left[
\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & \frac{1}{2} & \frac{1}{2} & 1 \\
0 & 2 & 1 & 5
\end{array}
\right]
$$

---

## 第二步：消去第二列下方的元素

**目标**：让第 3 行的第二列变成 0。

**变换操作**：
- $R_3 \leftarrow R_3 - 4R_2$

**变换后的矩阵（上三角形式）**：

$$
\left[
\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & \frac{1}{2} & \frac{1}{2} & 1 \\
0 & 0 & -1 & 1
\end{array}
\right]
$$

---

## 第三步：将主元化为 1

**变换操作**：
- $R_1 \leftarrow \frac{1}{2}R_1$
- $R_2 \leftarrow 2R_2$
- $R_3 \leftarrow -R_3$

**变换后的矩阵**：

$$
\left[
\begin{array}{ccc|c}
1 & \frac{1}{2} & -\frac{1}{2} & 4 \\
0 & 1 & 1 & 2 \\
0 & 0 & 1 & -1
\end{array}
\right]
$$

---

## 第四步：回代消元，化为行最简形（RREF）

**变换操作**：
- $R_2 \leftarrow R_2 - R_3$
- $R_1 \leftarrow R_1 + \frac{1}{2}R_3$
- $R_1 \leftarrow R_1 - \frac{1}{2}R_2$

**最终矩阵（行最简形）**：

$$
\left[
\begin{array}{ccc|c}
1 & 0 & 0 & 2 \\
0 & 1 & 0 & 3 \\
0 & 0 & 1 & -1
\end{array}
\right]
$$

---

## 最终解

$$
\begin{cases}
x = 2 \\
y = 3 \\
z = -1
\end{cases}
$$

> **提示**：在计算机实现中，通常只需消元到上三角矩阵，然后进行回代即可，不必完全化为单位矩阵。此外，为了数值稳定性，通常会引入列主元交换。

## 高斯消元（列主元法）

- $n \times n$ 矩阵 $A$，增广矩阵 $[A|b]$
- $k = 1$ 到 $n - 1$：
  1. **选主元**：在第 $k$ 列，找到绝对值最大的元素所在的行，与第 $k$ 行交换。
     - *目的：提高数值稳定性*
  2. **消元**：按你学过的基础消元即可

**结果**：得到上三角矩阵

## 解的三种情况：

- **唯一解**：矩阵的秩等于未知数个数 $n$
- **无穷多解**：矩阵的秩小于 $n$，且出现全零行（$0=0$）
- **无解**：矩阵的秩小于 $n$，且出现矛盾行（$0=\text{非零}$）
- 矩阵的秩，让我们知道方程组有多少“有效信息”。

### 高斯消元后的解判定

在得到上三角矩阵后，根据矩阵的秩和出现的特殊行，有三种情况：

| 情况 | 条件 | 说明 |
| :--- | :--- | :--- |
| 唯一解 | 秩 $= n$ | 主元齐全，可以直接回代求解 |
| 无穷多解 | 秩 $< n$ | 出现全零行（$0=0$），存在自由变量 |
| 无解 | 秩 $< n$ | 出现矛盾行（$0=\text{非零}$），方程组不相容 |

## 高斯消元 2.0： LU 分解

### 核心思想
如果我们把矩阵 $A$ 分解为 $A = LU$，其中：
- $L$ (Lower Traingular)：下三角矩阵，非零元素在下三角。
- $U$ (Upper Traingular)：上三角矩阵，非零元素在上三角。

$$
Ax = b
$$

1. 解 $Ly = b$ (前向代入，很快)
2. 解 $Ux = y$ (回代，很快)

### 与基础高斯消元的联系
基础高斯消元的目标就是把矩阵化为上三角矩阵 $U$，而在这个过程中所做的行变换，恰好可以由一个下三角矩阵 $L$ 表示。因此，$A = LU$ 本质上是**高斯消元的矩阵形式**。

### 为什么用 LU 分解？
如果遇到多个右端项 $b$ 需要求解（例如机器学习中多组标签），直接用高斯消元每次都要重复计算。而使用 $A = LU$ 后：
1. 先解 $Ly = b$（前代法，Forward Substitution）
2. 再解 $Ux = y$（回代法，Back Substitution）
这样只需分解一次，即可高效求解多个 $b$ 向量。

解方程组本质上就是找两条线的交点

## 病态方程组与数值稳定性

### 现象
矩阵 $A$ 的行列式接近 0，条件数极大。
几何上表现为方程组对应的直线几乎平行，交点极不稳定。

### 为什么在计算机中很致命？
计算机浮点数运算存在舍入误差（相当于对 $b$ 或 $A$ 施加微小扰动 $\delta$）。
在病态矩阵下，哪怕 $\delta$ 只有 $0.05$，解也可能从 $(1,1)$ 飞飞到 $(-4,6)$，导致计算结果完全不可信。

## 深度学习极度依赖归一化与正则化

深度学习本质上是求解一个极度复杂的非凸优化问题。由于网络很深、参数极多，损失曲面（Loss Landscape）经常表现出“病态”特征（类似于我们之前讨论的两条几乎平行的直线）。因此，DL 极度依赖归一化和正则化来维持数值稳定性。

### 1. 归一化 (Normalization)
**代表技术：Batch Normalization (BN), LayerNorm**
- **解决什么问题？** 深层网络在训练时，底层参数的更新会导致高层输入分布剧烈变化（称为“内部协变量偏移”）。这种变化会让高层网络不断适应新的数据分布，相当于一直在解一个条件数极大的病态方程。
- **数学本质**：BN 强制把每一层的输入重新拉回到均值 0、方差 1 的分布（$ \hat{x} = \frac{x - \mu}{\sigma} $）。
- **效果**：**重塑了损失曲面的几何形状**，让 Hessian 矩阵的特征值分布更均匀（降低条件数），从而允许使用更大的学习率，加速收敛，防止梯度爆炸或消失。

### 2. 正则化 (Regularization)
**代表技术：L2 正则化（权重衰减）、L1 正则化、Dropout**
- **解决什么问题？** 当参数数量远大于样本量时，模型极易过拟合。此时矩阵（如线性回归的 $X^T X$）接近奇异（行列式接近 0），求逆会“解飞走”，方差极大。
- **数学本质（以 L2 正则化为例）**：我们在损失函数中加上惩罚项 $\lambda ||w||^2$。这在数学上等价于给正规方程的对角线加一个正数：$ (X^T X + \lambda I) w = X^T y $。
- **效果**：**强行改善矩阵的条件数**。$\lambda I$ 的加入让原本接近 0 的特征值被“抬”了起来，使得矩阵可逆且数值稳定。这也就是为什么岭回归（Ridge Regression）能解决多重共线性问题。

### 核心总结
| 概念 | 对应线性代数中的动作 | 核心目的 |
| :--- | :--- | :--- |
| **归一化** | 调整数据分布，重塑特征空间 | 让 Hessian 矩阵条件数变小，优化更平滑 |
| **正则化** | 给矩阵对角线加 $\lambda I$ | 防止矩阵奇异，抑制过拟合，提高泛化 |
| **列主元消元** | 交换行，选最大主元 | 计算机求解时的数值稳定性（防除零） |

> 它们本质上都是在对抗“病态矩阵”带来的数值不稳定性。
