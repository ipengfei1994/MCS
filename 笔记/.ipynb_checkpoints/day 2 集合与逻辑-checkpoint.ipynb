````python
from unittest import result

users = [
    {'age': 20,'city':'Beijing'},
    {'age': 35,'city':'shanghai'},
    {'age': 65,'city':'shenzhen'},
]

#筛选age>18 and city == 'beijing'

result = [user for user in users if user['age'] > 18 and user['city'] =='Beijing']

print(result)
````

集合的源头：分类

- 第一步：把食物分成几堆
- 第二步：判断某个东西属不属于哪一堆（属于关系）
- 第三步：把两堆合在一起，或者从一堆里面拿走一些

以上是集合论的核心：分类、属于、运算

````python
import numpy as np
import matplotlib.pyplot as plt

# 设置中文字体，防止标题乱码
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

print('加载成功！')
````

````python
#逻辑运算运用
#程序判断就是逻辑运算
age = 20
has_id = True
````

````python
#p∩q：年龄满足且有证件才能进入
if age >= 18 and has_id:
    print('可以进入')
else:
    print('不可以进入')
````

````python
# 德摩根定律在代码中的应用
# ¬(p∩q)=¬p∪¬q

if not(age >= 18 and has_id):
    print('被拒绝')
````

````python
if age < 18 or not has_id:
    print('被拒绝')
````
