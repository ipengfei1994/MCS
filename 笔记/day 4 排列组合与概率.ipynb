排列组合俩原理

加法原理：如果做一件事情有m种方法，另一件n种，m+n

乘法原理：如果做一件事情有m种方法，另一件n种，m*n

排列：从n个不同元素中取r个，按照顺序排列

$$P(n, r) = \frac{n!}{(n-r)!}$$

组合：从n个不同元素中取r个，不考虑顺序：

$$C(n, r) = \frac{n!}{r!(n-r)!}$$

排列比组合多$r!$倍

古典概率

条件概率：  $P(A|B)$

独立事件：如果A和B独立，那么$P(A∩B)=P(A)P(B)$

二项分布：抛硬币、投色子、点击广告、检测疾病

$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}$$

几何分布：等待第一次成功、第一次中奖、第一次命中

$$P(X = k) = (1 - p)^{k-1} p$$

期望：随机变量的平均值


$$E[X] = \sum_{x} x P(X = x)$$

方差：随机变量偏离期望的程度

$$\operatorname{Var}(X) = E[(X - E[X])^2] = E[X^2] - (E[X])^2$$

二项分布的期望是np 方差np(1-p)

几何分布的期望是1/p 方差(1-p)/p^2

贝叶斯定理：

$$P(A \mid B) = \frac{P(B \mid A) P(A)}{P(B)}$$


P(A)：先验概率，你原先的信念

P(B|A)，如果A成立，B出现的概率

P(A|B):后验概率,看到B之后，你更新的信念


大数定律：

$$\frac{1}{n} \sum_{i=1}^{n} X_i \rightarrow E[X]$$

中心极限定理：统计学核心，解释了为什么正态分布无处不在

$$\frac{\sum X_i - n\mu}{\sigma \sqrt{n}} \rightarrow N(0,1)$$

```python
def factorial(n):
    r=1
    for i in range(2,n+1):r*=i
    return r

def P(n,r):
    r2=1
    for i in range(r):r2*=(n-i)
    return r2
def C(n,r):
    return  P(n,r)//factorial(r)

print('P(10, 3) =', P(10, 3))
print('C(10, 3) =', C(10, 3))
print('C(52, 5) =', C(52, 5))
```

```python
#作业1：扑克牌概率
from math import comb, factorial

def C(n, r):
    return comb(n, r)

def P(n, r):
    return factorial(n) // factorial(n - r)

# ============ 总组合数 ============
total = C(52, 5)
print("=" * 60)
print("作业 1：扑克牌 5 张牌型概率")
print("=" * 60)
print(f"总组合数 C(52, 5) = {total}")

# ============ 各种牌型组合数 ============
straight_flush = 4 * 10                                    # 4 花色 × 10 顺子
four_kind      = 13 * 48                                   # 13 点数选 4 张 × 48 张剩余
full_house     = 13 * C(4, 3) * 12 * C(4, 2)               # 13 选 3 张 × 12 选 2 张
flush          = 4 * C(13, 5) - straight_flush             # 4 花色各 5 张 - 同花顺
straight       = 10 * (4 ** 5) - straight_flush            # 10 顺子 × 4^5 花色 - 同花顺
three_kind     = 13 * C(4, 3) * C(12, 2) * 4 * 4           # 三条 + 两单张
two_pair       = C(13, 2) * C(4, 2) * C(4, 2) * 44         # 两对 + 单张
one_pair       = 13 * C(4, 2) * C(12, 3) * 4 ** 3          # 一对 + 三单张
high_card      = total - (straight_flush + four_kind +
                          full_house + flush + straight +
                          three_kind + two_pair + one_pair)  # 高牌

hands = {
    '同花顺': straight_flush,
    '四条':   four_kind,
    '葫芦':   full_house,
    '同花':   flush,
    '顺子':   straight,
    '三条':   three_kind,
    '两对':   two_pair,
    '一对':   one_pair,
    '高牌':   high_card,
}

# ============ 输出每种牌型概率 ============
print("-" * 60)
print(f"{'牌型':<8}{'组合数':>10}{'概率':>14}{'百分比':>14}")
print("-" * 60)
for name, count in hands.items():
    p = count / total
    print(f"{name:<8}{count:>10}{p:>14.6f}{p * 100:>13.4f}%")
print("-" * 60)

# ============ 结论 ============
print("\n【结论】")
print(f"1. 所有牌型概率之和 = {sum(hands.values()) / total:.2f}（应为 1）")
print(f"2. 最常见的牌型：{max(hands, key=hands.get)}（{max(hands.values()) / total * 100:.2f}%）")
print(f"3. 最罕见的牌型：{min(hands, key=hands.get)}（{min(hands.values()) / total * 100:.4f}%）")
print(f"4. 高牌 + 一对占比 = {(high_card + one_pair) / total * 100:.2f}%")
print(f"5. 大牌型（四条及以上）占比 = {(straight_flush + four_kind) / total * 100:.4f}%")
print(f"6. 验证分类完备：{sum(hands.values())} == {total} → {sum(hands.values()) == total}")
```

```python
#作业2：掷骰子：两个之和的分布
import random
import numpy as np
from collections import Counter

random.seed(42)
np.random.seed(42)

print("=" * 60)
print("作业 2：两个骰子之和的分布")
print("=" * 60)

# ============ 1. 理论分布 ============
outcomes = {}
for i in range(1, 7):
    for j in range(1, 7):
        s = i + j
        outcomes[s] = outcomes.get(s, 0) + 1

print("\n【理论分布】")
print(f"{'和':>4}{'组合数':>10}{'概率':>14}{'百分比':>14}")
print("-" * 60)
for s in range(2, 13):
    cnt = outcomes[s]
    print(f"{s:>4}{cnt:>10}{cnt / 36:>14.4f}{cnt / 36 * 100:>13.2f}%")

# ============ 2. 理论期望与方差 ============
theoretical_mean = sum(s * outcomes[s] / 36 for s in range(2, 13))
theoretical_var = sum((s - theoretical_mean) ** 2 * outcomes[s] / 36 for s in range(2, 13))

print("\n【理论统计量】")
print(f"E[S]     = {theoretical_mean:.4f}")
print(f"Var(S)   = {theoretical_var:.4f}")
print(f"σ(S)     = {np.sqrt(theoretical_var):.4f}")
print(f"验证：E[S] = 3.5 + 3.5 = {3.5 + 3.5}（两个独立骰子期望之和）")
print(f"验证：Var(S) = 35/12 + 35/12 = {35/12 + 35/12:.4f}（方差可加）")

# ============ 3. 模拟验证 ============
n_trials = 100000
sim_sums = [random.randint(1, 6) + random.randint(1, 6) for _ in range(n_trials)]
sim_counts = Counter(sim_sums)

print(f"\n【模拟分布（{n_trials:,} 次）】")
print(f"{'和':>4}{'模拟次数':>10}{'模拟概率':>14}{'理论概率':>14}")
print("-" * 60)
for s in range(2, 13):
    cnt = sim_counts.get(s, 0)
    print(f"{s:>4}{cnt:>10}{cnt / n_trials:>14.4f}{outcomes[s] / 36:>14.4f}")

sim_mean = np.mean(sim_sums)
sim_var = np.var(sim_sums)

print("\n【模拟统计量】")
print(f"模拟 E[S]   = {sim_mean:.4f}   （理论 {theoretical_mean:.4f}）")
print(f"模拟 Var(S) = {sim_var:.4f}   （理论 {theoretical_var:.4f}）")
print(f"模拟 σ(S)   = {np.sqrt(sim_var):.4f}   （理论 {np.sqrt(theoretical_var):.4f}）")

# ============ 4. 结论 ============
print("\n【结论】")
print(f"1. 和 = 7 概率最大：{outcomes[7]}/36 = {outcomes[7]/36*100:.2f}%（6 种组合方式）")
print(f"2. 和 = 2 或 12 概率最小：1/36 = {1/36*100:.2f}%（各只有 1 种）")
print(f"3. 分布呈三角形，关于 7 对称")
print(f"4. 期望 = 7，正好是两个骰子期望 3.5 + 3.5")
print(f"5. 方差 = 35/6 ≈ 5.833，等于两个骰子方差之和")
print(f"6. 模拟 10 万次后，模拟概率与理论概率误差 < 0.01")
print(f"7. 验证大数定律：模拟次数越多，频率越接近概率")

# ============ 5. 误差验证 ============
max_err = max(abs(sim_counts.get(s, 0) / n_trials - outcomes[s] / 36) for s in range(2, 13))
print(f"\n【误差验证】")
print(f"最大概率误差 = {max_err:.6f}")
print(f"是否所有误差 < 0.01：{max_err < 0.01}")
```
