## 特征方程

$Av = λv$<br>
λ为特征值 v特征向量<br>
矩阵A作用在v上相当于把v沿着原来的方向拉伸了λ倍<br>
$Av - λv = 0$ <br>
$(A - λ)v = 0$<br>
当且仅当系数矩阵的行列式为o:<br>
$det(A - λI)= 0$<br>

应用  λ和v的比较

### 稳定性分析（控制论）

动态系统是否稳定？取决于矩阵的特征值

如果所有特征值的实部都小于0，系统是稳定的（会衰减至0）<br>
如果所有特征值的实部都大于0，系统是发散的（爆炸）<br>

PCA为啥用协方差矩阵？协方差矩阵是对称的，特征向量正交，可以得到一组正交基，协方差矩阵的特征值反映了数据在其对应方向上的方差大小。<br>
特征值（EVD）的分解和SVD有什么区别？<br>
特征值可以为0嘛？为什么？答：完全可以，特征值为0意味着矩阵不可逆，行列式为0，对应的方向向量被压扁到了原点。<br>
特征向量一定要是单位向量嘛？不一定，特征向量可以任意缩放，只要保持方向不变即可。方向是特征向量的灵魂，长度只是它的“外衣”<br>
所有的矩阵都有特征值嘛？在实数范围内，不是所有的矩阵都有实特征值，但是对称矩阵有实特征值。<br>
特征值分解和svd有什么区别？EVD只适用于方阵，SVD可以适用于任意形状<br>
