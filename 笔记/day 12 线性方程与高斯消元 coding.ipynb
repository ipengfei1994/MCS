```python
#手写高斯消元加回代

def gauss_solve(A,b):
    n = len(b)
    M = np.hstack([A.astype(float).copy(),b.reshape(-1,1).astype(float)])
    for k in range(n):
        i = k +int(np.argmax(np.abs(M[k:,k])))
        M[[k,i]] =M[[i,k]]
        if abs(M[k,k])<1e-12:
            return None,M
        for i in range(k+1,n):
            f = M[i,k]/M[k,k]
            M[i]-=f*M[k]
    x = np.zeros(n)
    for i in range(n-1,-1,-1):
        x[i] = (M[i,-1]- M[i,i+1:n]@x[i+1:n]/M[i,i])
    return x,M

A = np.array([[2.0, 1.0, -1.0], [-3.0, -1.0, 2.0], [-2.0, 1.0, 2.0]])
b = np.array([8.0, -11.0, -3.0])
x, U = gauss_solve(A, b)
print('手写消元 x =', x)
print('numpy.solve x =', np.linalg.solve(A, b))
print('上三角增广矩阵')
print(np.round(U, 4))
```

```python
def diagnose(A,b):
    x,M = gauss_solve(A, b)
    print('增广矩阵消元后：')
    print(np.round(M, 4))
    if x is None:
        row = M[-1]
        if abs(row[-1])>1e-8:
            print('无解')
        else:
            print('无穷多解或需继续看秩')
    else:
        print('唯一解',x)

print('===看下相交===')
diagnose(np.array([[2., 1.], [1., 3.]]), np.array([5., 5.]))
print('===平行无解===')
diagnose(np.array([[1., 1.], [2., 2.]]), np.array([1., 3.]))
print('===重合无穷===')
diagnose(np.array([[1., 1.], [2., 2.]]), np.array([1., 2.]))
```

### 病态：扰动0.05

```python
A = np.array([[1.0,1.0],[1.0,1.01]])
b0 = np.array([2.0,2.01])
b1 = np.array([2.0,2.06])
x0 = np.linalg.solve(A,b0)
x1 = np.linalg.solve(A,b1)
print('原解',x0)
print('扰动后',x1)
print('解的改变量',np.linalg.norm(x1-x0))
print('cond(A) ≈',np.linalg.cond(A))
```

### 牛顿步就是解方程组

```python
xy = np.array([1.6,1.2])
g = np.array([2.0*xy[0],20.0*xy[1]])
H = np.array([[2.0,0.0],[0.0,20.0]])
d = np.linalg.solve(H,-g)

print('梯度 g=',g)
print('牛顿步 d=',d)
print('下一步 xy+d = ',xy+d,'(二次型应该到原点)')
print('错误示范：不要np.linalg.inv(H) @ (-g), 能 solve 就 solve！')
```
