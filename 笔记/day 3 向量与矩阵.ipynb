```python
#数组运算vs列表运算 计算速度
import time

size = 1000000
list1 = list(range(size))
list2 = list(range(size))

arr1 = np.array(list1)
arr2 = np.array(list2)

#测试列表运算
start_list = time.time()
list_result = [x + y for x, y in zip(list1, list2)]
end_list = time.time()
list_time = end_list - start_list
print(f"Python 列表运算耗时: {list_time:.5f} 秒")

#测试 NumPy 数组运算
start_numpy = time.time()
arr_result = arr1 + arr2
end_numpy = time.time()
numpy_time = end_numpy - start_numpy
print(f"NumPy 数组运算耗时: {numpy_time:.5f} 秒")

# 3. 计算并打印快了多少倍
speedup = list_time / max(numpy_time, 1e-8)
print(f"结论：NumPy 比 Python 列表快了约 {speedup:.1f} 倍！")
```

```python
#向量运算
v1=np.array([1,2,3])
v2=np.array([4,5,6])

print('v1+v2=',v1+v2)
print('v1*v2=',v1*v2)
print('v1v2',np.dot(v1,v2))
print('v1',(v1*v2).sum())
```

```python
norm_v1 = np.linalg.norm(v1)
print('|v1|',norm_v1)
```

```python
cos_theta = np.dot(v1,v2)/(np.linalg.norm(v1)*(np.linalg.norm(v2)))
angle = np.degrees(np.arccos(cos_theta))
print(f'夹角={angle:.1f}度')
```

```python
a = np.array([1,0])
b = np.array([0,1])

print('ab点积',np.dot(a,b),'-->垂直！')
```

```python
v = np.array([3,4])
unit_v = v /np.linalg.norm(v)
print('单位向量是：v=',unit_v,'长度是=',np.linalg.norm(unit_v))
```

```python
#矩阵变换的几何意义
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('A @ B =')
print(A @ B)
print('A * B =')
print(A * B)
```

```python
theta= np.pi/4
R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])

v = np.array([1,2])
v_rotated = R@v
print(f'原向量{v}旋转45度后={v_rotated}')
print(f'长度保持不变：|v|={np.linalg.norm(v):.3f},|Rv|={np.linalg.norm(v_rotated):.3f}')
```
